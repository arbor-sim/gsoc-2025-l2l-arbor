# Prototype: Arbor Allen Cell Fitting

Implementation of simple optimization using the validated allen-cell.ipynb method

- First run the model with Allen's original parameters to verify correctness
- Based on the successful implementation in allen-cell.ipynb ,optimize parameters knowing that a correct solution exists

```python
import arbor as A
from arbor import units as U
import numpy as np
import pandas as pd
import json
import matplotlib.pyplot as plt
from collections import defaultdict
from dataclasses import dataclass
from typing import Optional, Dict, List, Tuple
import time
from scipy.stats import qmc
```

## Load Allen Data

```python
# Load data
raw = A.load_swc_neuron('single_cell_allen.swc')
morphology = raw.morphology
labels = raw.labels
labels.add_swc_tags()

ref_data = pd.read_csv('single_cell_allen_neuron_ref.csv', usecols=[1,2], index_col=0)
with open('single_cell_allen_fit.json') as f:
    fit_data = json.load(f)

# Adjust reference values
ref_data['U/mV'] *= 1000.0
ref_data['U/mV'] += fit_data["fitting"][0]["junction_potential"]

# Extract reference spike times
threshold = -10
ref_voltage = ref_data['U/mV'].values
ref_time = ref_data.index.values
crossings = np.where((ref_voltage[:-1] < threshold) & (ref_voltage[1:] > threshold))[0]
ref_spike_times = ref_time[crossings + 1]
print(f"Reference spike count: {len(ref_spike_times)}")
print(f"Reference spike times: {ref_spike_times}")
```

## Parse Allen Parameters

```python
@dataclass
class parameters:
    cm:   Optional[float] = None # uF/cm2
    temp: Optional[float] = None # K
    Vm:   Optional[float] = None # mV
    rL:   Optional[float] = None # Ohm cm

# Get default parameters
default = parameters(
    temp=float(fit_data["conditions"][0]["celsius"]) * U.Celsius,
    Vm=float(fit_data["conditions"][0]["v_init"]) * U.mV,
    rL=float(fit_data["passive"][0]["ra"]) * U.Ohm * U.cm
)

# Parse parameters
param = defaultdict(parameters)
mechs = defaultdict(dict)

for block in fit_data["genome"]:
    mech = block["mechanism"] or "pas"
    region = block["section"]
    name = block["name"]
    value = float(block["value"])
    
    if name.endswith("_" + mech):
        name = name.removesuffix("_" + mech)
    elif mech == "pas":
        if name == "cm":
            param[region].cm = value * U.uF / U.cm2
        elif name == "Ra":
            param[region].rL = value * U.Ohm * U.cm
        elif name == "Vm":
            param[region].Vm = value * U.mV
        elif name == "celsius":
            param[region].temp = value * U.Celsius
        else:
            raise Exception(f"Unknown key: {name}")
        continue
    else:
        raise Exception(f"Illegal combination {mech} {name}")
    
    mechs[(region, mech)][name] = value

# Get ion reversal potentials
erev = dict()
for kv in fit_data["conditions"][0]["erev"]:
    region = kv["section"]
    for k, v in kv.items():
        if k == "section":
            continue
        ion = k[1:]
        erev[(region, ion)] = float(v)*U.mV
```

## Define AllenCellModel

```python
class AllenCellModel:
    def __init__(self, morphology, labels, default_params, region_params, 
                 mechanisms, erev, stimulus_current=150*U.pA):
        self.morphology = morphology
        self.labels = labels
        self.default_params = default_params
        self.region_params = region_params
        self.mechanisms = mechanisms
        self.erev = erev
        self.stimulus_current = stimulus_current
        # Fixed discretization policy
        self.cvp = A.cv_policy_max_extent(20)
        
    def create_cell(self, params):
        """Create cell based on successful allen-cell.ipynb method"""
        dec = A.decor()
        
        # Set default properties
        dec.set_property(
            tempK=self.default_params.temp,
            Vm=self.default_params.Vm,
            cm=self.default_params.cm,
            rL=self.default_params.rL,
        )
        
        # Override region parameters
        for region, vs in self.region_params.items():
            dec.paint(
                f'"{region}"',
                tempK=vs.temp,
                Vm=vs.Vm,
                cm=vs.cm,
                rL=vs.rL,
            )
        
        # Apply mechanisms - using allen-cell.ipynb method
        for (region, mech), values in self.mechanisms.items():
            nm = mech
            vs = {}
            sp = "/"
            
            # Update parameter values
            for k, v in values.items():
                if mech == "pas" and k == "e":
                    nm = f"{nm}{sp}{k}={v}"
                    sp = ","
                elif mech == "pas" and k == "g":
                    # Handle passive conductance optimization
                    param_key = f"g_pas_{region}"
                    if param_key in params:
                        vs[k] = params[param_key]
                    else:
                        vs[k] = v
                else:
                    # Handle other mechanism parameters
                    param_key = f"{k}_{mech}_{region}"
                    if param_key in params:
                        vs[k] = params[param_key]
                    else:
                        vs[k] = v
            
            # Use A.mechanism wrapper
            dec.paint(f'"{region}"', A.density(A.mechanism(nm, vs)))
        
        # Add stimulus and threshold detector
        dec.place('(location 0 0.5)', A.iclamp(200*U.ms, 1000*U.ms, self.stimulus_current), "ic")
        dec.place('(location 0 0.5)', A.threshold_detector(-10*U.mV), "dt")
        
        # Set ion reversal potentials
        for (region, ion), e in self.erev.items():
            dec.paint(f'"{region}"', ion=ion, rev_pot=e)
        
        # Set calcium Nernst equation
        dec.set_ion("ca", int_con=5e-5*U.mM, ext_con=2.0*U.mM, method="nernst/x=ca")
        
        # Create cell
        return A.cable_cell(self.morphology, dec, self.labels, self.cvp)
    
    def simulate(self, params, tfinal=1400*U.ms, dt=0.005*U.ms):
        """Run simulation and return results"""
        cell = self.create_cell(params)
        
        # Use single_cell_model method (verified to correctly detect spikes)
        model = A.single_cell_model(cell)
        model.properties.catalogue.extend(A.allen_catalogue(), '')
        
        # Set probe
        model.probe('voltage', '(location 0 0.5)', 'Um', frequency=200 * U.kHz)
        
        # Run simulation
        model.run(tfinal=tfinal, dt=dt)
        
        # Get results
        voltage_trace = None
        if model.traces:
            time = model.traces[0].time
            voltage = model.traces[0].value
            voltage_trace = np.column_stack((time, voltage))
        
        spike_times = list(model.spikes)
        
        return voltage_trace, spike_times
```

### Optimization Parameter Selection Rationale

Based on the successful implementation in `allen-cell.ipynb` and the complete Allen database parameter set, we selected the following parameters for optimization:

1. **Sodium Channel Conductances (gbar_NaV)**:
   - All regions (soma, axon, dend, apic)
   - Directly control action potential generation and propagation
   - Most critical parameters for spike count and timing

2. **Kv3.1 Potassium Channel Conductances (gbar_Kv3_1)**:
   - Expressed in all regions
   - Fast-activating delayed rectifier potassium channels
   - Control action potential repolarization speed and firing frequency

3. **Passive Conductances (g_pas)**:
   - Determine resting membrane potential and input resistance
   - Affect subthreshold membrane dynamics
   - Work with e_pas to set resting state

4. **Calcium Channel Conductances (gbar_Ca_HVA, gbar_Ca_LVA)**:
   - Mainly expressed in soma and axon
   - Trigger calcium-activated potassium channels
   - Affect spike afterhyperpolarization

5. **Calcium Dynamics Parameters (gamma, decay)**:
   - Control intracellular calcium concentration change rates
   - Affect time course of calcium-dependent processes

Parameter range setting principles:
- Default values near range center
- Range width allows ±20-50% variation

## Define Fitness Function

```python
def compute_fitness(model, params, ref_spike_times, ref_voltage, ref_time):
    """Calculate fitness between model output and reference data"""
    try:
        voltage_trace, spike_times = model.simulate(params)
        
        # 1. Spike count error
        spike_count_error = abs(len(spike_times) - len(ref_spike_times))
        
        # 2. Spike timing error
        if len(spike_times) > 0 and len(ref_spike_times) > 0:
            # Match nearest spikes
            spike_time_errors = []
            for ref_spike in ref_spike_times:
                if len(spike_times) > 0:
                    min_error = min(abs(spike - ref_spike) for spike in spike_times)
                    spike_time_errors.append(min_error)
                else:
                    spike_time_errors.append(100)  # Penalty for missing spike
            spike_time_error = np.mean(spike_time_errors)
        else:
            spike_time_error = 100
        
        # 3. Voltage trace error (optional)
        if voltage_trace is not None:
            # Interpolate to reference time points
            sim_voltage = np.interp(ref_time, voltage_trace[:, 0], voltage_trace[:, 1])
            voltage_error = np.sqrt(np.mean((sim_voltage - ref_voltage)**2))
        else:
            voltage_error = 1000
        
        # Combined fitness
        fitness = (
            10 * spike_count_error +  # Spike count most important
            spike_time_error +        # Spike timing secondary
            0.01 * voltage_error      # Voltage trace lower weight
        )
        
        return fitness, {
            'spike_count': len(spike_times),
            'spike_count_error': spike_count_error,
            'spike_time_error': spike_time_error,
            'voltage_error': voltage_error
        }
        
    except Exception as e:
        print(f"Simulation error: {e}")
        return 10000, None
```

## Simple Optimization Algorithm

```python
def random_search_optimization(model, param_space, ref_spike_times, ref_voltage, ref_time, 
                             n_iterations=50, n_samples_per_iter=10):
    """Pure random search optimization - samples completely randomly from parameter ranges each time"""
    
    # First calculate fitness of default parameters as target
    default_params = {k: v['default'] for k, v in param_space.items()}
    default_fitness, default_info = compute_fitness(model, default_params, ref_spike_times, ref_voltage, ref_time)
    print(f"Default parameters (Allen original) fitness: {default_fitness:.2f}")
    if default_info:
        print(f"  Default spike count: {default_info['spike_count']}, target: {len(ref_spike_times)}")
        print(f"  Default spike timing error: {default_info['spike_time_error']:.2f} ms")
    
    print(f"\nGoal: Find fitness <= {default_fitness:.2f}")
    print("=" * 60)
    
    # Initialize best parameters (random)
    best_params = {}
    for param, info in param_space.items():
        best_params[param] = np.random.uniform(info['min'], info['max'])
    
    best_fitness = float('inf')
    history = []
    
    # Pure random search
    total_evaluations = 0
    for iteration in range(n_iterations):
        
        # Generate multiple completely random parameter combinations each iteration
        for sample in range(n_samples_per_iter):
            # Generate parameters completely randomly
            test_params = {}
            for param, info in param_space.items():
                test_params[param] = np.random.uniform(info['min'], info['max'])
            
            # Evaluate
            fitness, info = compute_fitness(model, test_params, ref_spike_times, ref_voltage, ref_time)
            total_evaluations += 1
            
            # Update best parameters
            if fitness < best_fitness:
                best_fitness = fitness
                best_params = test_params.copy()
                
                print(f"\nIteration {iteration+1} (evaluation {total_evaluations}): New best fitness = {fitness:.2f}")
                if info:
                    print(f"  Spike count: {info['spike_count']}, target: {len(ref_spike_times)}")
                    print(f"  Spike timing error: {info['spike_time_error']:.2f} ms")
                
                # Check if target reached
                if best_fitness <= default_fitness:
                    print(f"\n✅ Success! Found parameters as good as or better than defaults")
                    print(f"  Final fitness: {best_fitness:.2f} <= Target: {default_fitness:.2f}")
                    print(f"  Total evaluations: {total_evaluations}")
                    history.append(best_fitness)
                    return best_params, best_fitness, history
        
        history.append(best_fitness)
        
        # Print progress every 10 iterations
        if (iteration + 1) % 10 == 0:
            print(f"\nProgress: {iteration + 1}/{n_iterations} iterations complete")
            print(f"  Current best fitness: {best_fitness:.2f}")
            print(f"  Total evaluations: {total_evaluations}")
    
    print(f"\nOptimization complete.")
    print(f"  Total evaluations: {total_evaluations}")
    print(f"  Final fitness: {best_fitness:.2f}")
    if best_fitness > default_fitness:
        print(f"  ⚠️ Did not reach default parameter performance")
        print(f"  Gap: {best_fitness - default_fitness:.2f}")
    
    return best_params, best_fitness, history
```

## Ground Truth Validation

First run the model with Allen's original parameters to verify our implementation correctly reproduces the reference data

```python
# Create model
model = AllenCellModel(
    morphology=morphology,
    labels=labels,
    default_params=default,
    region_params=param,
    mechanisms=mechs,
    erev=erev,
    stimulus_current=150*U.pA
)

# Use Allen original parameters (default values)
ground_truth_params = {k: v['default'] for k, v in optimization_params.items()}

print("=" * 80)
print("Ground Truth Validation - Using Allen Original Parameters")
print("=" * 80)

# Run ground truth simulation
gt_voltage_trace, gt_spike_times = model.simulate(ground_truth_params)
gt_fitness, gt_info = compute_fitness(model, ground_truth_params, ref_spike_times, ref_voltage, ref_time)

print(f"\nGround Truth Results:")
print(f"  Model spike count: {len(gt_spike_times)}")
print(f"  Reference spike count: {len(ref_spike_times)}")
print(f"  Model spike times: {gt_spike_times}")
print(f"  Reference spike times: {ref_spike_times}")
print(f"  Fitness: {gt_fitness:.2f}")

# Plot ground truth comparison
fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(gt_voltage_trace[:, 0], gt_voltage_trace[:, 1], 'b-', linewidth=2, 
        label=f'Ground Truth (Allen parameters) - {len(gt_spike_times)} spikes')
ax.plot(ref_time, ref_voltage, 'r--', linewidth=2, alpha=0.7, 
        label=f'Reference data - {len(ref_spike_times)} spikes')

# Mark spikes
for spike_time in gt_spike_times:
    ax.axvline(spike_time, color='blue', linestyle=':', alpha=0.5)
for spike_time in ref_spike_times:
    ax.axvline(spike_time, color='red', linestyle=':', alpha=0.5)

ax.set_xlabel('Time (ms)')
ax.set_ylabel('Membrane Potential (mV)')
ax.set_title('Ground Truth Validation: Allen Original Parameters vs Reference Data')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 1400)
plt.tight_layout()
plt.show()

# Check if optimization is needed
if len(gt_spike_times) == len(ref_spike_times) and gt_fitness < 5.0:
    print("\n✅ Ground Truth is already very close to reference data!")
else:
    print("\n⚠️ Ground Truth differs from reference data")
```

## Run Optimization

Now that we know a correct solution exists, start the optimization process

```python
# Run optimization
print("\n" + "=" * 80)
print("Starting Parameter Optimization...")
print("Goal: Starting from random parameters, find performance equivalent to Allen originals")
print("=" * 80)
start_time = time.time()

best_params, best_fitness, history = random_search_optimization(
    model, 
    optimization_params, 
    ref_spike_times,
    ref_voltage,
    ref_time,
    n_iterations=50,
    n_samples_per_iter=10
)

print(f"\nOptimization complete! Time: {time.time() - start_time:.1f} seconds")
print(f"\nFinal Results Comparison:")
print(f"  Ground truth (Allen original) fitness: {gt_fitness:.2f}")
print(f"  Optimized parameters fitness: {best_fitness:.2f}")

if best_fitness <= gt_fitness:
    print(f"\n✅ Success! The optimization algorithm found parameters as good as or better than Allen originals")

else:
    print(f"\nOptimization did not reach Allen original performance")
```
